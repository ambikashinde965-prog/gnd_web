```python
from functools import reduce

numbers = [10, 25, 5, 40, 15]

maximum = reduce(lambda x, y: x if x > y else y, numbers)
minumum = reduce(lambda x, y: x if x < y else y, numbers)

print("Maximum:", maximum)
print("Minumum:",minumum)
```
